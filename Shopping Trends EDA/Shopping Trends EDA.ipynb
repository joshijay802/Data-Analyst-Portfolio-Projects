```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('shopping_trends.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df["Age"].plot(kind = "hist", edgecolor = "black")
plt.title("Age Distribution Histogram")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.countplot(data=df, x="Gender", hue="Gender")
plt.title("Gender Distribution of Shoppers")
plt.xlabel("Gender")
plt.ylabel("Number of Shoppers")
plt.show()
```

```python
df.groupby('Category')['Purchase Amount (USD)'].mean().plot(kind = "bar")
plt.title('Average Purchase Amount by Category')
plt.xlabel('Categories')
plt.ylabel('Average Purchase Amount (USD)')
plt.show()
```

```python
df["Subscription Status"].value_counts().plot(kind = "pie", autopct='%1.1f%%')
plt.title("Subscription Status Distribution")
plt.show()
```

```python
df.groupby("Season")["Purchase Amount (USD)"].sum().plot(kind = "bar")
plt.title("Total Purchase Amount by Season")
plt.xlabel("Season")
plt.ylabel("Total Purchase Amount (USD)")
plt.show()
```

```python
df["Season"].value_counts().plot(kind = "pie", autopct='%1.1f%%')
plt.title('Sales Distribution among Seasons')
plt.show()
```

```python
plt.figure(figsize=(15,8))
df.groupby("Location")["Purchase Amount (USD)"].mean().sort_values(ascending = False).plot(kind = "bar")
plt.title('Purchase Amount by Location')
plt.xlabel('Location')
plt.ylabel('Total Purchase Amount (USD)')
plt.show()
```

```python
df.head()
df.groupby("Category")["Review Rating"].mean().plot(kind = "bar")
plt.title("Average Review Rating by Category")
plt.xlabel("Category")
plt.ylabel("Average Review Rating")
plt.show()
```

```python
df.head()
df.boxplot(column= "Purchase Amount (USD)", by= "Frequency of Purchases", figsize=(10,8))
plt.title("Purchase Amounts by Frequency of Purchases")
plt.suptitle("Distribution of Purchase Amounts", fontsize=8)
plt.xlabel("Frequency of Purchases")
plt.ylabel("Purchase Amount (USD)")
plt.show()
```

```python
df.head()
plt.figure(figsize=(20,8))
sns.histplot( data = df, x = "Location", bins=20, kde=True)
plt.title("Distribution of Locations")
plt.xlabel("Location")
plt.ylabel("Frequency")
# plt.grid(True)
plt.tight_layout()
plt.xticks( rotation=90)
plt.show()
```
